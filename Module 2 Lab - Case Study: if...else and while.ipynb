```python

# Name: Sabrina Keener
# Filename: student_gpa.py
# Description: This program asks for a student's name and GPA and determines
# whether the student qualifies for the Dean's List or Honor Roll.

while True:
    last_name = input("Enter the student's last name (or ZZZ to quit): ")

    if last_name.upper() == "ZZZ":
        break

    first_name = input("Enter the student's first name: ")
    gpa = float(input("Enter the student's GPA: "))

    if gpa >= 3.5:
        print(first_name, last_name, "has made the Dean's List.")
    elif gpa >= 3.25:
        print(first_name, last_name, "has made the Honor Roll.")
    else:
        print(first_name, last_name, "has not made the Dean's List or Honor Roll.")

    print()
```
